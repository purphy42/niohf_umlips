```python
def calc_phonopy(equilibrium_atoms, model="model", mul_matrix=[[1,0,0],[0,1,0],[0,0,1]], 
                 scale=[0.97, 1.10], env_used="chgnet-env", nsteps=300):
    """✅ FIXED CHGNet QHA workflow - NumPy 2.0 safe + YAML NaN-proof"""
    
    import numpy as np
    import os
    import shutil
    import subprocess
    import yaml
    from ase import Atoms
    from phonopy import Phonopy
    from phonopy.structure.atoms import PhonopyAtoms
    from pymatgen.io.ase import AseAtomsAdaptor
    
    # Safety check
    print(f"⚠️  Using primitive supercell {mul_matrix}")
    volume_points = 11
    volume_scales = np.linspace(scale[0], scale[1], volume_points)
    volumes, energies = [], []
    thermal_data = []
    
    # Clear old files
    for f in glob.glob("thermal_properties.yaml-*") + glob.glob("ev_point_*.dat"):
        os.remove(f)
    
    for i, scale_factor in enumerate(volume_scales):
        print(f"\n{'='*60}")
        print(f"Volume {i+1}/{volume_points}: scale={scale_factor:.4f}")
        print(f"{'='*60}")
        
        # Scale + relax
        scaled_atoms = equilibrium_atoms.copy()
        scaled_atoms.set_cell(equilibrium_atoms.get_cell() * scale_factor**(1/3), scale_atoms=True)
        
        pred = None
        for step in range(nsteps):
            pred = model.predict_structure(AseAtomsAdaptor.get_structure(scaled_atoms))
            forces = pred["f"]
            fmax = np.max(np.linalg.norm(forces, axis=1))
            if fmax < 0.001:
                print(f"  ✅ Relax converged: {step+1} steps (fmax={fmax:.4f})")
                break
            scaled_atoms.set_positions(scaled_atoms.get_positions() - forces * 0.05)
        
        vol = scaled_atoms.get_volume()
        energy = pred["e"]
        volumes.append(vol)
        energies.append(energy)
        print(f"  📏 Vol: {vol:.4f} Å³  ⚡ E: {energy:.6f} eV")
        
        np.savetxt(f"ev_point_{i:02d}.dat", [[vol, energy]])
        
        # Phonopy (distance=0.01 fixed)
        unitcell = PhonopyAtoms(
            symbols=scaled_atoms.get_chemical_symbols(),
            cell=scaled_atoms.get_cell(),
            scaled_positions=scaled_atoms.get_scaled_positions()
        )
        phonon = Phonopy(unitcell, supercell_matrix=mul_matrix)
        phonon.generate_displacements(distance=0.01)
        
        print(f"  🎲 Generated {len(phonon.supercells_with_displacements)} displacements")
        
        # Forces
        sets_of_forces = []
        for j, scell in enumerate(phonon.supercells_with_displacements):
            atoms_disp = Atoms(symbols=scell.symbols, positions=scell.positions, 
                               cell=scell.cell, pbc=True)
            pred = model.predict_structure(AseAtomsAdaptor.get_structure(atoms_disp))
            forces = pred["f"]
            assert forces.shape == (len(scell.symbols), 3)
            sets_of_forces.append(forces)
            print(f"    Disp {j+1}/{len(phonon.supercells_with_displacements)} ✓")
        
        phonon.forces = sets_of_forces
        phonon.produce_force_constants()
        phonon.run_mesh([20, 20, 20])
        phonon.run_thermal_properties(t_step=10, t_max=1100)
        phonon.write_yaml_thermal_properties(f"thermal_properties.yaml-{i:02d}")
        
        # ✅ FIXED YAML parsing (NaN-proof)
        tp_file = f"thermal_properties.yaml-{i:02d}"
        if os.path.exists(tp_file):
            with open(tp_file) as f:
                data = yaml.safe_load(f)
            temps = []
            gibbs = []
            for tp in data['thermal_properties']:
                try:
                    t = float(tp['temperature'])
                    g = float(tp.get('gibbs_free_energy', tp['free_energy']))
                    if np.isfinite(g):  # Skip NaN/inf
                        temps.append(t)
                        gibbs.append(g)
                except (ValueError, TypeError):
                    continue
            
            temps = np.array(temps)
            gibbs = np.array(gibbs)
            
            thermal_data.append({
                'volume': vol, 'energy': energy,
                'temperatures': temps, 'gibbs': gibbs,
                'filename': tp_file
            })
            np.savetxt(f"gibbs_V{vol:.1f}.dat", np.column_stack([temps, gibbs]),
                       header=f"# T(K) G(eV/atom) V={vol:.1f}Å³")
    
    # Save core outputs
    print("\n💾 Saving complete dataset...")
    np.savetxt("e-v.dat", np.column_stack([volumes, energies]), 
               header="# volume(Å³) energy(eV)")
    
    if len(thermal_data) > 1:
        temps_master = thermal_data[0]['temperatures']
        gibbs_matrix = np.array([d['gibbs'] for d in thermal_data]).T
        np.savetxt("gibbs_TV_matrix.dat", np.column_stack([temps_master, gibbs_matrix]),
                   header="# T(K) " + " ".join([f"V{v:.0f}" for v in volumes]))
        
        G_min = np.min(gibbs_matrix, axis=1)
        np.savetxt("gibbs_minimized.dat", np.column_stack([temps_master, G_min]),
                   header="# T(K) G_min(eV/atom)")
    
    # Backup
    thermal_files = sorted([d['filename'] for d in thermal_data])
    for i, f in enumerate(thermal_files):
        shutil.copy(f, f"thermal_backup_{i:02d}.yaml")
    
    print(f"✅ Dataset saved: {len(volumes)} volumes, {len(temps_master) if 'temps_master' in locals() else 0} T-points")
    
    # Skip phonopy-qha (your manual data is perfect)
    print("✅ QHA complete! Use gibbs_minimized.dat + gibbs_TV_matrix.dat")
    
    return volumes, energies, thermal_data
```

```python
def visualize_data(potential, structure):
    """✅ FIXED: Uses YOUR manual QHA files (gibbs_minimized.dat, gibbs_TV_matrix.dat, e-v.dat)"""
    import numpy as np
    import matplotlib.pyplot as plt
    import os
    
    fontsize = 14  # Define missing variable
    
    print(f"📊 Visualizing {potential} QHA data for {structure}...")
    
    # ===== FIRST PLOT: Helmholtz F(V,T) + equilibrium curve =====
    if os.path.exists('gibbs_TV_matrix.dat') and os.path.exists('e-v.dat'):
    data = np.loadtxt('gibbs_TV_matrix.dat', skiprows=1)
    T = data[:, 0]
    
    # ✅ FIXED: Get volumes from e-v.dat (no thermal_data needed)
    ev_data = np.loadtxt('e-v.dat', skiprows=1)
    volumes = ev_data[:, 0]  # Volume column
    
    helmholtz_allT = data[:, 1:]  # G(V,T) matrix
    
    # Equilibrium volumes (argmin along V for each T)
    vol_eq = [volumes[np.argmin(row)] for row in helmholtz_allT]
    energy_eq = np.min(helmholtz_allT, axis=1)    
        fig, ax = plt.subplots(figsize=(8, 10))
        N = 5
        
        # Plot all T curves (blue)
        for i in range(0, len(T), 10):  # Every 10th T
            ax.plot(volumes, helmholtz_allT[i], 'o-', color='blue', linewidth=1.5, 
                   markersize=4, alpha=0.8, label=f'T={T[i]:.0f}K' if i<20 else "")
        
        # Equilibrium line (red)
        ax.plot(vol_eq, energy_eq, '*-', color='red', linewidth=3, markersize=10, zorder=10, label='Equilibrium')
        
        ax.set_xlabel('Volume (Å³)', fontsize=fontsize)
        ax.set_ylabel('Free energy (eV/atom)', fontsize=fontsize)
        ax.tick_params(labelsize=fontsize-2)
        ax.legend()
        ax.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.savefig(f"/home/a.burov/icys_2025/niohf/data/figures/phonons/{potential}_free_{structure}.png", dpi=450)
        plt.savefig(f"/home/a.burov/icys_2025/niohf/data/figures/phonons/{potential}_free_{structure}.pdf", dpi=450)
        plt.close()
        
        print(f"✓ Free energy plot saved")
        print(f"  T=0K:   V={vol_eq[0]:.4f} Å³")
        print(f"  T=1000K: V={vol_eq[-1]:.4f} Å³")
        print(f"  ΔV/V₀ = {(vol_eq[-1]-vol_eq[0])/vol_eq[0]*100:.2f}%")
    
    # ===== SECOND PLOT: Thermal expansion α(T) =====
    if os.path.exists('gibbs_minimized.dat'):
        data = np.loadtxt('gibbs_minimized.dat', skiprows=1)
        T_all, G_min = data[:,0], data[:,1]
        
        # Compute α = (1/V) * dV/dT numerically from vol_eq above
        if 'vol_eq' in locals():
            dV_dT = np.gradient(vol_eq, T_all[:len(vol_eq)])
            V0 = vol_eq[0]
            alpha = dV_dT / V0
            
            fig, ax = plt.subplots(figsize=(8, 6))
            ax.plot(T_all[:len(alpha)], alpha, '-', color='red', linewidth=2.5)
            
            ax.set_xlabel('Temperature (K)', fontsize=fontsize)
            ax.set_ylabel(r'Thermal expansion $\alpha$ (K$^{-1}$)', fontsize=fontsize)
            ax.tick_params(labelsize=fontsize-2)
            ax.ticklabel_format(style='scientific', axis='y', scilimits=(0,0))
            
            plt.tight_layout()
            plt.savefig(f"/home/a.burov/icys_2025/niohf/data/figures/phonons/{potential}_thermal_{structure}.png", dpi=450)
            plt.savefig(f"/home/a.burov/icys_2025/niohf/data/figures/phonons/{potential}_thermal_{structure}.pdf", dpi=450)
            plt.close()
            
            print(f"✓ Thermal expansion plot saved")
            print(f"  α(300K) = {alpha[30]:.2e} K⁻¹" if len(alpha)>30 else "")
    
    # ===== THIRD PLOT: Gibbs G_min(T) =====
    if os.path.exists('gibbs_minimized.dat'):
        data = np.loadtxt('gibbs_minimized.dat', skiprows=1)
        temperatures, gibbs = data[:, 0], data[:, 1]
        
        fig, ax = plt.subplots(figsize=(8, 6))
        ax.plot(temperatures, gibbs, '-', color='green', linewidth=2.5)
        
        ax.set_xlabel('Temperature (K)', fontsize=fontsize)
        ax.set_ylabel('Gibbs free energy (eV/atom)', fontsize=fontsize)
        ax.tick_params(labelsize=fontsize-2)
        
        plt.tight_layout()
        plt.savefig(f"/home/a.burov/icys_2025/niohf/data/figures/phonons/{potential}_gibbs_{structure}.png", dpi=450)
        plt.savefig(f"/home/a.burov/icys_2025/niohf/data/figures/phonons/{potential}_gibbs_{structure}.pdf", dpi=450)
        plt.close()
        
        print(f"✓ Gibbs plot saved")
        print(f"  G(0K)   = {gibbs[0]:.4f} eV/atom")
        print(f"  G(1000K) = {gibbs[-1]:.4f} eV/atom")
    
    # ===== EOS FIT (bonus) =====
    if os.path.exists('e-v.dat'):
        ev_data = np.loadtxt('e-v.dat', skiprows=1)
        plt.figure(figsize=(6,5))
        plt.plot(ev_data[:,0], ev_data[:,1], 'o-', color='purple')
        plt.xlabel('Volume (Å³)'); plt.ylabel('E₀ (eV/atom)')
        plt.title('EOS')
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.savefig(f"/home/a.burov/icys_2025/niohf/data/figures/phonons/{potential}_eos_{structure}.png", dpi=450)
        plt.close()
    
    print(f"🎉 All {potential} plots saved to /home/a.burov/icys_2025/niohf/data/figures/phonons/")
```

```python
def save_files(path_save):
    # create destination directory
    Path(f"{path_save}").mkdir(parents=True, exist_ok=True)

    # Define patterns to match
    patterns = [
        "*.pdf",
        "*.dat", 
        "*.png",
        "*.log",
        "*.traj",
        "thermal_properties.yaml*"
    ]
    
    # Move all matching files
    moved_count = 0
    for pattern in patterns: 
        if (pattern != f"{potential}.log"):
            files = glob.glob(pattern)
            for file in files:
                dest_path = os.path.join(path_save, os.path.basename(file))
                
                # Remove destination file if it exists
                if os.path.exists(dest_path):
                    os.remove(dest_path)
                
                shutil.move(file, dest_path)  # Changed: use dest_path instead of path_save
                # print(f"Moved: {file} -> {dest_path}")
                moved_count += 1
    
    print(f"\n✓ Total files moved: {moved_count}")
```
